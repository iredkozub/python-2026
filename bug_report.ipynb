#### Задание 2 (обязательно). Свой баг-репорт
Самостоятельно придумать и оформить один новый баг на ту же тему (изменяемость/alias/hashable), которого не было в каталоге семинара. Файл hw3/my_bug.py должен содержать три части подряд:

1. багованный код (5-10 строк);
2. комментарий в 2-3 предложения: что происходит и почему;
3. исправленная версия.

Запрещается просто переформулировать чужой пример с семинара другими именами переменных - цель в том, чтобы придумать свою ситуацию, а не скопировать готовый.

```python
store_a = [100, 200, 300]
store_b = [150, 250]

def combine_sales(first, second):
    result = first
    result += second
    return result

combined = combine_sales(store_a, store_b)

print("Combined:", combined)
print("Store A:", store_a)
print("Store B:", store_b)
```

Функция объединяет списки продаж двух магазинов, но в исходной реализации изменяет первый список. Причина в том, что result = first не создаёт копию, а связывает два имени с одним объектом, поэтому операция result += second изменяет исходные данные. Для исправления ошибки используется first.copy(), позволяющий объединить списки без изменения оригинала

```python
import copy

store_a = [100, 200, 300]
store_b = [150, 250]

def combine_sales(first, second):
    result = first.copy()
    result += second
    return result

combined = combine_sales(store_a, store_b)

print("Combined:", combined)
print("Store A:", store_a)
print("Store B:", store_b)
```
